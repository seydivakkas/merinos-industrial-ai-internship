### 1. Problem
Bir endüstriyel üretim işletmesinde (tekstil/halı sanayi) bilgisayar mühendisliği problemleri tek bir kalıba sığmaz. Yazılıma doğrudan başlamadan önce firmanın üretim akışını, çalışma istasyonlarının hesaplama kapasitelerini ve ortaya çıkan heterojen veri kaynaklarını (sayısal sensör telemetrisi, yüksek çözünürlüklü halı yüzey fotoğrafları, teknik bakım dokümanları) doğru kategorize etmek gerekir.

### 2. Why the Problem Matters
Verinin doğasını (yapılandırılmış, yarı yapılandırılmış, yapılandırılmamış) ve donanım sınırlarını (CPU, RAM, depolama) analiz etmeden seçilen kütüphaneler ve algoritmalar üretim ortamında başarısız olur. Örneğin görüntü işleme piksel matrisleri gerektirirken, telemetri zaman serisi analizi, dokümanlar ise metinsel NLP yöntemleri gerektirir.

### 3. Engineering Concepts
- **Veri Modaliteleri (Modalities)**: Sayısal (Numerical), Görsel (Visual), Metinsel (Textual).
- **Sistem Profili (Environment Profiling)**: İşlemci çekirdek sayısı, bellek (RAM) limitleri ve Python çalışma zamanı uyumluluğu.
- **Yerel PoC Sınırları**: Saha operasyonunu riske atmadan sentetik ve örnek verilerle güvenli yerel geliştirme.

```python
# 4. Library / API Investigation
import os
import sys
import platform
import psutil
from pydantic import BaseModel

print(f"Python Version: {sys.version}")
print(f"OS: {platform.system()} {platform.release()}")
print(f"CPU Logical Cores: {os.cpu_count()}")
print(f"RAM: {psutil.virtual_memory().total / (1024**3):.2f} GB")
```

```python
# 5. Minimal Implementation
from day01.mini_project.src.models import DataAsset, DataModality, SystemResourceRequirements
from day01.mini_project.src.environment_profiler import EnvironmentProfiler

reqs = SystemResourceRequirements(
    min_cpu_cores=2,
    min_ram_gb=4.0,
    required_python_version="3.10"
)
profiler = EnvironmentProfiler(requirements=reqs)
audit = profiler.audit_workstation()
print("Workstation Audit:", audit.model_dump_json(indent=2))
```

```python
# 6. Experiment: Enterprise Data Assets Inventory
assets = [
    DataAsset(asset_id="NUM-001", name="Weaving Loom Tension", modality=DataModality.NUMERICAL, source_type="LoomSensor", estimated_size_kb=50.0),
    DataAsset(asset_id="VIS-001", name="Carpet Jacquard Pattern", modality=DataModality.VISUAL, source_type="OpticalCamera", estimated_size_kb=4500.0),
    DataAsset(asset_id="TXT-001", name="SOP Weft Maintenance Manual", modality=DataModality.TEXTUAL, source_type="PDFManual", estimated_size_kb=1200.0),
]
for a in assets:
    profiler.register_asset(a)

dist = profiler.get_modality_distribution()
print("Modality Distribution:", dist)
```

```python
# 7. Visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.bar(dist.keys(), dist.values(), color=["#2b5c8f", "#d95f02", "#7570b3"])
plt.title("Tekstil İşletmesinde Veri Türleri Dağılımı (PoC Envanteri)")
plt.xlabel("Veri Modalitesi")
plt.ylabel("Varlık Sayısı")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 8. Validation
assert audit.is_compliant, "Sistem donanım gereksinimlerini karşılamıyor!"
assert sum(dist.values()) == 3, "Envanter varlık sayısı uyuşmuyor!"
print("Tüm sistem ve envanter doğrulamaları başarılı.")
```
