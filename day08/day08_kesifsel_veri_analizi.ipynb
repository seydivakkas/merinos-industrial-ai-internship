### 1. Problem
Modelleme aşamasına geçmeden önce verinin dağılımını, çarpıklığını (skewness), değişkenler arasındaki gizli doğrusal ilişkileri ve sensör arızalarından kaynaklanan ekstrem uç değerleri (outliers) analiz etmek gerekir.

### 2. Why the Problem Matters
Keşifsel Veri Analizi (EDA), veri setindeki gizli örüntüleri ortaya çıkarır. Aykırı değerler tespit edilmeden eğitilen modeller ciddi genelleme hataları yapar.

### 3. Engineering Concepts
- **IQR (Interquartile Range)**: Çeyrekler açıklığı tabanlı aykırı değer eşiği: $[Q_1 - 1.5\cdot IQR, Q_3 + 1.5\cdot IQR]$.
- **Pearson Korelasyonu**: İki sürekli değişken arasındaki doğrusal ilişki katsayısı ($[-1, 1]$).
- **Histogram & Kutu Grafiği (Boxplot)**: Değişken dağılımının ve medyan etrafındaki yayılımın görselleştirilmesi.

```python
# 4. Library / API Investigation
import pandas as pd
import numpy as np
from day08.mini_project.src.eda_toolkit import EDAToolkit

toolkit = EDAToolkit()
print("EDA Araç Seti Hazır.")
```

```python
# 5. Minimal Implementation
np.random.seed(42)
n = 100
df = pd.DataFrame({
    "motor_temp_c": np.concatenate([np.random.normal(75, 4, n-3), [135, 142, 128]]),
    "pneumatic_pressure_bar": np.random.normal(14.5, 0.8, n),
    "rpm": np.random.normal(820, 25, n)
})

report = toolkit.analyze(df)
print(f"Satır Sayısı: {report.row_count} | Sütun Sayısı: {report.column_count}")
print("Sıcaklık İstatistikleri:", report.columns["motor_temp_c"].model_dump())
```

```python
# 6. Experiment: Korelasyon Matrisi ve Aykırı Değerler
print("En Güçlü İlişkili Değişken Çiftleri:")
for c1, c2, score in report.top_correlated_pairs:
    print(f"  {c1} <-> {c2}: |r| = {score:.4f}")
```

```python
# 7. Visualization: Dağılım ve Aykırı Değerler
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(9, 4))
axes[0].hist(df["motor_temp_c"], bins=20, color="#3182bd", edgecolor="black")
axes[0].set_title("Motor Sıcaklığı Dağılımı")
axes[0].set_xlabel("Sıcaklık (°C)")
axes[0].set_ylabel("Frekans")

axes[1].boxplot(df["motor_temp_c"], vert=True, patch_artist=True, boxprops=dict(facecolor="#9ecae1"))
axes[1].set_title("Sıcaklık Kutu Grafiği (Aykırı Değerler)")
plt.tight_layout()
plt.show()
```

```python
# 8. Validation
assert report.columns["motor_temp_c"].outlier_count_iqr >= 3
print("Aykırı değerler IQR kuralı ile başarıyla yakalandı.")
```

```python
# 9. Failure Cases: Sayısal Olmayan Veri
try:
    toolkit.analyze(pd.DataFrame({"text": ["a", "b"]}))
except ValueError as e:
    print("Beklenen hata yakalandı:", e)
```
