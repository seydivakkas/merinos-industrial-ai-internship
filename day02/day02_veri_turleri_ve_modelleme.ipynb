### 1. Problem
Endüstriyel bir halı fabrikasında ürün bilgileri (en, boy, desen adı), üretim fotoğrafları ve teknik notlar farklı veri biçimlerinde saklanır. Bir ürünün sabit nitelikleri tabloda tutulabilirken, her ürüne ait değişken sayıda yüksek çözünürlüklü görselin ve etiketlerin tek bir düz tabloda tutulması veri tekrarına ve tutarsızlığa yol açar.

### 2. Why the Problem Matters
Doğru veri modeli seçilmediğinde sistemde veri anomalileri (referential integrity bozulması, eksik görseller, ilişkisiz kayıtlar) oluşur. CSV düz satır/sütun için basit ve hızlıyken, hiyerarşik JSON iç içe ilişkilerde esneklik sağlar.

### 3. Engineering Concepts
- **Yapılandırılmış Veri (Structured)**: Sabit şemalı, tip tanımlı tablo (CSV, SQL).
- **Yarı Yapılandırılmış Veri (Semi-Structured)**: Anahtar-değer veya belge odaklı esnek yapı (JSON, NoSQL).
- **Yapılandırılmamış Veri (Unstructured)**: Piksel dizileri, ham metin veya ses.
- **Referential Integrity (Referans Bütünlüğü)**: Ürün ID ile Görsel ID arasındaki ilişki tutarlılığı.

```python
# 4. Library / API Investigation & Minimal Implementation
from typing import Dict, List, Optional, Tuple
from pydantic import BaseModel, Field

class ProductTabularRecord(BaseModel):
    product_id: str
    title: str
    collection: str
    width_cm: float
    length_cm: float
    primary_color: str

class VisualAssetMetadata(BaseModel):
    image_id: str
    file_name: str
    angle_or_view: str
    resolution: str

class ProductCompositeCatalog(BaseModel):
    product_id: str
    title: str
    collection: str
    dimensions: Dict[str, float]
    primary_color: str
    visual_assets: List[VisualAssetMetadata] = []

class SchemaTransformer:
    def __init__(self):
        self.products: Dict[str, ProductTabularRecord] = {}
        self.assets: Dict[str, VisualAssetMetadata] = {}
        self.links: Dict[str, List[str]] = {}

    def add_product(self, p: ProductTabularRecord):
        self.products[p.product_id] = p

    def add_asset(self, a: VisualAssetMetadata):
        self.assets[a.image_id] = a

    def link_product_to_images(self, pid: str, img_ids: List[str]):
        self.links[pid] = img_ids

    def validate_referential_integrity(self) -> Tuple[bool, List[str]]:
        errs = []
        for pid, img_ids in self.links.items():
            if pid not in self.products:
                errs.append(f"Product {pid} not found")
            for iid in img_ids:
                if iid not in self.assets:
                    errs.append(f"Asset {iid} referenced by product {pid} not found")
        return len(errs) == 0, errs

    def build_composite_catalog(self, pid: str) -> ProductCompositeCatalog:
        p = self.products[pid]
        linked_imgs = [self.assets[iid] for iid in self.links.get(pid, []) if iid in self.assets]
        return ProductCompositeCatalog(
            product_id=p.product_id,
            title=p.title,
            collection=p.collection,
            dimensions={"width_cm": p.width_cm, "length_cm": p.length_cm},
            primary_color=p.primary_color,
            visual_assets=linked_imgs
        )

print("Modeller basariyla tanimlandi.")
```

```python
# 5. Minimal Implementation
transformer = SchemaTransformer()
prod = ProductTabularRecord(product_id="PRD-01", title="Isparta Vintage", collection="Classic", width_cm=160, length_cm=230, primary_color="Bordo")
img = VisualAssetMetadata(image_id="IMG-01", file_name="isparta_top.jpg", angle_or_view="top_down", resolution="1920x1080")
transformer.add_product(prod)
transformer.add_asset(img)
transformer.link_product_to_images("PRD-01", ["IMG-01"])

valid, errs = transformer.validate_referential_integrity()
print(f"Referential Integrity: {valid} (Hatalar: {errs})")
```

```python
# 6. Experiment: Composite JSON Transformation
composite = transformer.build_composite_catalog("PRD-01")
print("Composite JSON Document:")
print(composite.model_dump_json(indent=2))
```

```python
# 7. Visualization
import matplotlib.pyplot as plt

formats = ["Structured (CSV)", "Semi-Structured (JSON)", "Unstructured (Image/Text)"]
flexibility = [3, 8, 10]
query_simplicity = [9, 7, 2]

plt.figure(figsize=(7, 4))
x = range(len(formats))
plt.bar([p - 0.2 for p in x], flexibility, width=0.4, label="Şema Esnekliği", color="#4575b4")
plt.bar([p + 0.2 for p in x], query_simplicity, width=0.4, label="Basit Tablo Sorgulanabilirliği", color="#d73027")
plt.xticks(x, formats)
plt.ylabel("Puan (1-10)")
plt.title("Veri Formatları Karşılaştırmalı Mühendislik Analizi")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 8. Validation
assert composite.product_id == "PRD-01"
assert len(composite.visual_assets) == 1
assert composite.dimensions["width_cm"] == 160
print("Tüm şema ve ilişki testleri başarıyla doğrulandı.")
```

```python
# 9. Failure Cases: Missing Image Reference
transformer.link_product_to_images("PRD-01", ["NON_EXISTENT_IMAGE"])
is_valid, errs = transformer.validate_referential_integrity()
print("Beklenen ilişki hatası yakalandı:", errs)
```
